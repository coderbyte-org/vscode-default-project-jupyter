## 2. Work with data

Calculate summary statistics for a sample dataset.

```python
from statistics import mean, median

scores = [72, 85, 91, 68, 95, 88]

print(f"Average: {mean(scores):.2f}")
print(f"Median: {median(scores)}")
print(f"Highest score: {max(scores)}")
```

## 3. Visualize data

Create a simple text-based bar chart using Python.

```python
for i, score in enumerate(scores, start=1):
  print(f"Student {i}: {"#" * (score // 5)} {score}")
```
